## Part B — The physics and climate forcing

Along a one-dimensional flowline, conservation of ice gives

$$\frac{\partial H}{\partial t}=\dot b-\frac{\partial q}{\partial x}.$$

- $H$: glacier thickness (m); $\dot b$: **surface mass balance (SMB)** (m/year); $q$: ice flux (m²/year).
- In the **shallow-ice approximation (SIA)**, gravity-driven flux depends strongly on ice thickness and ice-surface slope.
- A simplified climate rule makes SMB positive above the ELA and negative below it:

$$\dot b(S)=\min\left[0.001(S-z_{ELA}),\ 0.3\right]\quad (\mathrm{m/year}).$$

**Run the next cell to plot this climate rule yourself.** It compares the starting ELA of **3,000 m** with an ELA raised to **3,500 m**. Change `ELA_after` or `test_elevation`, then rerun. Where does snow accumulation become net ice loss after the ELA rises?

```python
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams.update({"figure.dpi": 110, "axes.spines.top": False, "axes.spines.right": False})
print("Ready: NumPy", np.__version__)

# Part B: PLOT the surface mass balance
ELA_after = 3500
test_elevation = 3200  # EDIT: try 2800, 3200, or 3800 m

ELA_before = 3000
surface_elevations = np.linspace(2300, 4600, 500)  # elevation (m)

def smb_at_elevation(elevation, ela):
    # Identical formula to the glacier solver in Part D (m ice / year)
    return np.minimum(0.001 * (np.asarray(elevation) - ela), 0.3)

smb_before = smb_at_elevation(surface_elevations, ELA_before)
smb_after = smb_at_elevation(surface_elevations, ELA_after)

fig, ax = plt.subplots(figsize=(9, 4.7))
ax.axhline(0, color="0.25", lw=1)
ax.plot(surface_elevations, smb_before, color="#0B8499", lw=3,
        label="Original ELA = 3000 m")
ax.plot(surface_elevations, smb_after, color="#E07833", lw=3,
        label=f"Raised ELA = {ELA_after} m")
ax.axvline(ELA_before, color="#0B8499", lw=1.5, ls="--", alpha=.6)
ax.axvline(ELA_after, color="#E07833", lw=1.5, ls="--", alpha=.6)
ax.axvline(test_elevation, color="0.45", ls=":", lw=1.5)

before_here = float(smb_at_elevation(test_elevation, ELA_before))
after_here = float(smb_at_elevation(test_elevation, ELA_after))
ax.scatter([test_elevation, test_elevation], [before_here, after_here],
           c=["#0B8499", "#E07833"], s=55, zorder=4, edgecolors="white")
ax.set(xlim=(2300, 4600), ylim=(-1.3, .45), xlabel="Ice-surface elevation (m)",
       ylabel="Surface mass balance (m ice/year)",
       title="How does raising the ELA change accumulation and ablation?")
ax.text(2350, .34, "Positive SMB: ice gain", color="#0B8499", fontsize=10)
ax.text(2350, -1.18, "Negative SMB: ice loss", color="#C0612D", fontsize=10)
ax.grid(alpha=.18)
ax.legend(loc="lower right", fontsize=9)
plt.tight_layout()
plt.show()
print(f"At {test_elevation:.0f} m, original SMB = {before_here:+.2f} m/year; "
      f"with ELA {ELA_after} m, SMB = {after_here:+.2f} m/year.")
print("QUESTION: Which elevations switch from net ice gain to net ice loss?")
```

## Part C — ice thickness and ice surface slope controls flow speed

Run this small Glen-law/SIA example. **Change `H` from 50 to 100 m.** What happens to speed near the surface? Here we hold slope and ice softness fixed, assume no basal sliding, and calculate a vertical velocity profile.

```python
A = 1.0e-16      # Pa^-3 year^-1, ice softness
rho_g = 910 * 9.81
slope_deg = 24
H = 50.0         # EDIT: try 100 m
z = np.linspace(0, H, 120)
slope = -np.tan(np.deg2rad(slope_deg))
u = -0.5 * A * (rho_g * slope)**3 * (H**4 - (H-z)**4)
plt.figure(figsize=(6,4))
plt.plot(u, z, linewidth=3)
plt.xlabel("Speed (m/year)");plt.ylabel("Height above glacier bed (m)")
plt.title(f"Velocity profile for {H:.0f} m of ice")
plt.grid(alpha=.25);plt.show()
print(f"Surface speed ≈ {u[-1]:.1f} m/year")
```

## Part D — Watch a glacier grow toward steady state

**Run both code cells below in order:**

1. **Model definitions:** The first cell contains the SIA solver. It defines functions and **does not plot by itself**. You do not need to edit it.
2. **▶ Watch the glacier grow:** The next cell actually runs the model from an ice-free landscape and displays an animation with playback controls.

We use a 140 km idealized flowline with 700 m spacing. The glacier grows under a constant **ELA of 3,000 m** for **6,000 model years**, long enough for this particular experiment to get **close to steady state**. Notice how the ice-surface profile, glacier length, and cross-sectional area gradually stop changing and plateau.

The SIA model assumes **no basal sliding, simple elevation-dependent mass balance, and fixed bedrock**. The cross-sectional area (km²) is a **proxy for ice volume** in a 2D geometry.


```python
# SIA flowline model: provided code (no editing required)
import numpy as np

L = 140_000.0       # m, idealized model domain
N = 201             # grid points (200 intervals, 700 m spacing)
x = np.linspace(0.0, L, N)
dx = x[1] - x[0]
bed = 13522.8 - 1000.0 * np.log(x + 1000.0)  # m, idealized sloping terrain
rho_g = 910.0 * 9.81 # Pa / m
A_DEFAULT = 2.5e-15 # Pa^-3 year^-1 (Glen exponent n=3)
SMB_GRAD = 0.001    # (m ice / year) per meter elevation
SMB_CAP = 0.3       # m ice / year, maximum annual accumulation

def ice_velocity_profile(thickness=50.0, slope_degrees=23.0, A=1.0e-16):
    """SIA no-sliding depth-profile for a simple uniform slab, units m/yr."""
    z = np.linspace(0, float(thickness), 150)
    slope = -np.tan(np.deg2rad(slope_degrees))
    u = -0.5*A*(rho_g*slope)**3 * (thickness**4-(thickness-z)**4)
    return z, u

def glacier_length_km(H, min_thickness=5.0):
    present = np.flatnonzero(H > min_thickness)
    return float(x[present[-1]]/1000.0) if len(present) else 0.0

def integrate(initial_H, years, ELA=3000.0, A=A_DEFAULT, save_interval=50.0):
    """Finite-volume style, explicit 1-D SIA evolution with adaptive stable dt.

    Conserves flux in interior cells. Boundary thickness fixed at zero;
    negative thickness is clipped to zero after SMB and flux updates.
    Returns sampled thickness and model diagnostics.
    """
    H = np.asarray(initial_H, dtype=float).copy()
    assert H.shape == (N,)
    H[0] = 0.0; H[-1] = 0.0
    t = 0.0
    constant = 2.0/5.0 * A * rho_g**3
    snapshots = [(0.0, H.copy())]
    next_save = float(save_interval)
    while t < years-1e-10:
        S = bed + H
        slope = np.diff(S) / dx
        Hmid = (H[:-1]+H[1:]) / 2.0
        D = constant*Hmid**5*slope**2
        q = -D*slope
        dt = min(1.0, dx**2/(8*np.max(D)+1e-10), years-t)
        smb = np.minimum(SMB_GRAD * (S-ELA), SMB_CAP)
        H[1:-1] += dt * (-np.diff(q)/dx + smb[1:-1])
        H = np.maximum(H,0.0)
        H[0] = H[-1] = 0.0
        t += dt
        if t >= next_save - 1e-8:
            snapshots.append((t,H.copy()))
            next_save += save_interval
    if abs(t-snapshots[-1][0]) > 1e-6:
        snapshots.append((t,H.copy()))
    times = np.array([v[0] for v in snapshots])
    profiles = np.stack([v[1] for v in snapshots])
    lengths = np.array([glacier_length_km(v) for v in profiles])
    volumes = np.sum(0.5*(profiles[:,:-1]+profiles[:,1:]), axis=1)*dx/1e6 # 2D cross-section (km^2)
    return {"time":times,"H":profiles,"length_km":lengths,
            "area_km2":volumes,"final_H":H,"ELA":ELA,"A":A}

def experiment(ela_after=3500.0, ice_softness=A_DEFAULT, years_after=1000.0):
    """Spin up near steady state under ELA 3000 for 6000 years, then diverge."""
    initial = np.zeros(N)
    reference = integrate(initial,6000.0,ELA=3000.0,A=ice_softness,save_interval=100)
    H0 = reference['final_H']
    baseline = integrate(H0,years_after,ELA=3000.0,A=ice_softness,save_interval=25)
    warmed = integrate(H0,years_after,ELA=ela_after,A=ice_softness,save_interval=25)
    return reference, baseline, warmed
```

```python
# Part D: RUN this cell to see the glacier grow! (No editing required)
# It produces a playable animation directly below the code cell.
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

spin = integrate(np.zeros(N), years=6000, ELA=3000, save_interval=100)
km = x / 1000.0

fig, axes = plt.subplot_mosaic(
    [["profile", "profile"], ["length", "area"]],
    figsize=(8.8, 6.4), dpi=120,
    gridspec_kw={"height_ratios": [2.3, 1.0]}
)
# Use fixed margins, not tight_layout(), so changing animation text cannot
# shift the plot or crop titles when Matplotlib encodes individual frames.
fig.subplots_adjust(left=.11, right=.97, bottom=.12, top=.78,
                    hspace=.58, wspace=.34)
fig.suptitle("Part D - Glacier grows toward a steady state",
             fontsize=12, weight="bold", y=.97)
spin_status = fig.text(.5, .905, "Year 0 / 6000  |  Glacier length 0.0 km",
                       ha="center", va="center", fontsize=10)
ax_geom, ax_length, ax_area = axes["profile"], axes["length"], axes["area"]

ax_geom.plot(km, bed, color="0.35", lw=1.5, label="Bedrock")
ax_geom.axhline(3000, ls="--", color="#B06A3F", lw=1.4, label="ELA = 3000 m")
(ice_surface,) = ax_geom.plot([], [], lw=2.8, color="#087CA7", label="Ice surface")
ice_shading = [ax_geom.fill_between(km, bed, bed, color="#7BC4E6", alpha=.55)]
ax_geom.set(xlim=(0, 110), ylim=(1600, 6800),
            xlabel="Distance along glacier (km)", ylabel="Elevation (m)")
ax_geom.grid(alpha=.16)
ax_geom.legend(loc="upper right", fontsize=9)

# Full curves appear gradually as model time advances.
(line_length,) = ax_length.plot([], [], color="#087CA7", lw=2.5)
(dot_length,) = ax_length.plot([], [], "o", color="#087CA7", ms=7)
ax_length.set(xlim=(0, 6000), ylim=(0, max(spin["length_km"])*1.12 + 1),
              xlabel="Model year", ylabel="Glacier length (km)",
              title="Length approaches a plateau")
ax_length.grid(alpha=.18)

(line_area,) = ax_area.plot([], [], color="#087CA7", lw=2.5)
(dot_area,) = ax_area.plot([], [], "o", color="#087CA7", ms=7)
ax_area.set(xlim=(0, 6000), ylim=(0, max(spin["area_km2"])*1.12),
            xlabel="Model year", ylabel="Cross-sectional area (km²)",
            title="Area approaches a plateau")
ax_area.grid(alpha=.18)

def draw_spin_frame(i):
    ice = spin["H"][i]
    ice_surface.set_data(km, np.where(ice > 5, bed + ice, np.nan))
    ice_shading[0].remove()
    ice_shading[0] = ax_geom.fill_between(km, bed, bed + ice,
                                         color="#7BC4E6", alpha=.55)
    yr = spin["time"][i]
    line_length.set_data(spin["time"][:i+1], spin["length_km"][:i+1])
    dot_length.set_data([yr], [spin["length_km"][i]])
    line_area.set_data(spin["time"][:i+1], spin["area_km2"][:i+1])
    dot_area.set_data([yr], [spin["area_km2"][i]])
    spin_status.set_text(f"Year {yr:.0f} / 6000  |  "
                         f"Glacier length {spin['length_km'][i]:.1f} km")

animation_spin = FuncAnimation(fig, draw_spin_frame,
                               frames=len(spin["time"]), interval=130,
                               repeat=False, blit=False)
# to_jshtml embeds self-contained playback controls; works in Colab and Jupyter.
spin_movie = HTML(animation_spin.to_jshtml(fps=7, default_mode="once"))
plt.close(fig)            # Prevent an extra blank/static Matplotlib figure.
display(spin_movie)

last_1000 = 100 * (spin["area_km2"][-1] - spin["area_km2"][-11]) / spin["area_km2"][-1]
#print(f"After 6000 years: length = {spin['length_km'][-1]:.1f} km; "
      #f"cross-sectional area = {spin['area_km2'][-1]:.2f} km².")
#print(f"Area change over the final 1000 years: {last_1000:+.3f}% "
      #"(close to steady state, not exactly stationary).")
```

## Part E — Your experiment: raise the ELA and watch the response

You have a glacier that is now **in steady state**. What if the climate suddenly changes?

**Run the cell below.** It starts two glaciers from the exact same initial condition:

- **Constant climate** (blue/teal): ELA stays at 3,000 m.
- **Warming climate** (orange): ELA jumps to **3,500 m** (500 m higher) at year 0 and stays there.

The animation shows **both ice surfaces, changing glacier length, and changing cross-sectional area** over the next 1,000 years.

**Then change one number** in the code: `ELA_warm = 3300`, `3500`, or `3700` m, and rerun. What happens to the pace and magnitude of retreat? Why does the glacier continue to adjust after the ELA has already stopped changing?

```python
# Part E: RUN the experiment, then press ▶ to watch the transient response.
ELA_warm = 3500   # EDIT THIS NUMBER: try 3300, 3500, or 3700 m
spin, control, warm = experiment(ela_after=ELA_warm, years_after=1000)
km = x / 1000.0

fig, axes = plt.subplot_mosaic(
    [["profile", "profile"], ["length", "area"]],
    figsize=(8.8, 6.4), dpi=120,
    gridspec_kw={"height_ratios": [2.3, 1.0]}
)
# Fixed layout keeps the entire header visible in every animation frame.
fig.subplots_adjust(left=.11, right=.97, bottom=.12, top=.78,
                    hspace=.58, wspace=.34)
fig.suptitle("Part E - Glacier response to a higher ELA (warming climate)",
             fontsize=12, weight="bold", y=.97)
warm_status = fig.text(.5, .905, "Year 0 / 1000  |  retreat 0.0 km",
                       ha="center", va="center", fontsize=10)
ax_geom, ax_length, ax_area = axes["profile"], axes["length"], axes["area"]
ax_geom.plot(km, bed, color="0.35", lw=1.5, label="Bedrock")
ax_geom.axhline(3000, color="#0B8499", lw=1.2, ls="--", alpha=.7,
                label="Original ELA (3000 m)")
ax_geom.axhline(ELA_warm, color="#E07833", lw=1.2, ls="--", alpha=.7,
                label=f"New ELA ({ELA_warm} m)")
(control_surf,) = ax_geom.plot([], [], color="#0B8499", lw=2.8,
                               label="Constant ELA")
(warm_surf,) = ax_geom.plot([], [], color="#E07833", lw=2.8,
                            label="Raised ELA")
ice_shading = [ax_geom.fill_between(km, bed, bed, color="#0B8499", alpha=.12),
               ax_geom.fill_between(km, bed, bed, color="#E07833", alpha=.22)]
ax_geom.set(xlim=(0, 110), ylim=(1600, 6800),
            xlabel="Distance along glacier (km)", ylabel="Elevation (m)")
ax_geom.grid(alpha=.16)
ax_geom.legend(loc="upper right", fontsize=8)

ax_length.set(xlim=(0, 1000),
              ylim=(0, max(control["length_km"].max(), warm["length_km"].max())*1.12),
              xlabel="Years since ELA change", ylabel="Glacier length (km)",
              title="Terminus response")
ax_area.set(xlim=(0, 1000),
            ylim=(0, max(control["area_km2"].max(), warm["area_km2"].max())*1.12),
            xlabel="Years since ELA change", ylabel="Cross-sectional area (km²)",
            title="Ice volume response")
for ax in (ax_length, ax_area):
    ax.grid(alpha=.18)

length_lines = [ax_length.plot([], [], color=c, lw=2.4)[0]
                for c in ("#0B8499", "#E07833")]
length_dots = [ax_length.plot([], [], "o", color=c, ms=6)[0]
               for c in ("#0B8499", "#E07833")]
area_lines = [ax_area.plot([], [], color=c, lw=2.4)[0]
              for c in ("#0B8499", "#E07833")]
area_dots = [ax_area.plot([], [], "o", color=c, ms=6)[0]
             for c in ("#0B8499", "#E07833")]

def draw_warming_frame(i):
    yr = warm["time"][i]
    for j, result in enumerate((control, warm)):
        ice = result["H"][i]
        (control_surf if j == 0 else warm_surf).set_data(
            km, np.where(ice > 5, bed + ice, np.nan))
        ice_shading[j].remove()
        ice_shading[j] = ax_geom.fill_between(
            km, bed, bed + ice, color=("#0B8499" if j == 0 else "#E07833"),
            alpha=(.12 if j == 0 else .22))
        length_lines[j].set_data(result["time"][:i+1], result["length_km"][:i+1])
        length_dots[j].set_data([result["time"][i]], [result["length_km"][i]])
        area_lines[j].set_data(result["time"][:i+1], result["area_km2"][:i+1])
        area_dots[j].set_data([result["time"][i]], [result["area_km2"][i]])
    warm_status.set_text(f"Year {yr:.0f} / 1000  |  "
                         f" Retreat: {control['length_km'][i] - warm['length_km'][i]:.1f} km")

animation_warm = FuncAnimation(fig, draw_warming_frame,
                               frames=len(warm["time"]), interval=140,
                               repeat=False, blit=False)
warming_movie = HTML(animation_warm.to_jshtml(fps=7, default_mode="once"))
plt.close(fig)
display(warming_movie)

#print(f"Starting length: {spin['length_km'][-1]:.1f} km")
#print(f"After 1000 y, unchanged ELA: {control['length_km'][-1]:.1f} km")
#print(f"After 1000 y, raised ELA to {ELA_warm} m: {warm['length_km'][-1]:.1f} km")
#print(f"Ice cross-sectional area: {control['area_km2'][-1]:.2f} km² (unchanged) "
#     f"versus {warm['area_km2'][-1]:.2f} km² (raised ELA)")
```
